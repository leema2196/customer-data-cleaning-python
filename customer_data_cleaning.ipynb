```python
import pandas as pd
import numpy as np
print("\nLibraries are imported")
```

```python
data={'customer_id':[101,102,103,104,105,106,107,108,109,110,111,112,113,114,115],
      'name':['Rahul','Priya','Arun','Sneha','Rahul','Kiran','Meena','John','Divya','Ravi','Anu','Kumar','Ravi','Sara','Vijay'],
      'age':[25,28,np.nan,30,25,27,29,np.nan,32,26,31,28,26,24,35],
      'city':['Bangalore','Chennai','Bangalore','Mumbai','Bangalore',np.nan,'Delhi','Chennai','Mumbai','Delhi','Bangalore','Chennai','Delhi','Mumbai','Bangalore'],
      'salary':[30000,40000,35000,50000,30000,45000,np.nan,30000,55000,32000,42000,48000,32000,28000,60000],
      'email':['rahul@gmail.com','priya@gmail.com','arun@gmail.com','sneha@gmail.com','rahul@gmail.com','kiran@gmail.com','meena@gmail.com','john@gmail.com','divya@gmail.com','ravi@gmail.com','anu@gmail.com','kumar@gmail.com','ravi@gmail.com','sara@gmail.com','vijay@gmail.com']}
df=pd.DataFrame(data)
print('\n',df)
```

```python
print('\nDataset Information')
print(df.info())

print('\nDataset Shape')
print(df.shape)

print('\nMissing Values')
print(df.isnull().sum())


print('Total Records',len(df))
print('\nDuplicate Values',df.duplicated(subset='email').sum())

print(df[df.duplicated()])
```

```python
df=df.drop_duplicates(subset='email')
print('\nAfter deleting the duplicate records')
print(df)
print('\ntotal records after removing the duplicate records',len(df))
```

```python
print('age before filling missing values')
print(df['age'])
df['age']=df['age'].fillna(df['age'].median())
print('\n\nage after filling missing values')
print(df['age'])

df['salary']=df['salary'].fillna(df['salary'].median())
print(df['salary'])


df['city']=df['city'].fillna(df['city'].mode()[0])
print(df['city'])


print('\n',df.isnull().sum())
```

```python
df['name']=df['name'].str.title()
df['city']=df['city'].str.title()
df['email']=df['email'].str.lower()
print(df.head())
```

```python
print("total customers",len(df))
print("average age : ")
print(round(df['age'].mean(),2))
print("average salary : ",round(df['salary'].mean(),2))
print("salary summary : ")
print(df['salary'].describe())
print("customers by city : ")
print(df['city'].value_counts())
```

```python
df.to_excel('cleaned_customer_data.xlsx',index=False)
print("excel report generated successfully")
```
